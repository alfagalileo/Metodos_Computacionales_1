## Programación orientada a objetos

## Bucles
Son estructuras de control que nos permiten ejecutar un bloque de código varias veces. En python existen dos tipos de bucles:

1. **Bucles for**: Se ejecutan un número determinado de veces.
2. **Bucles while**: Se ejecutan mientras se cumpla una condición.

### Bucles for

Estos bucles se utilizan cuando conocemos el número de iteraciones que vamos a realizar. Para ejecutar un bucle for en python, necesitamos una estructura iterable (como una lista, tupla, diccionario, string, etc.) y una variable que vaya cambiando en cada iteración.

```python
for variable in estructura_iterable:
    # código a ejecutar
```

Ejemplo:
```python
for i in range(5):
    print(i)
```

La instrucción `range(5)` crea una secuencia de números del 0 al 4, y la variable `i` toma cada uno de esos valores en cada iteración.
En general, la instrucción `for in` se utiliza para recorrer elementos de una estructura iterable.

Ejemplo con una lista:
```python
lista = [1, 2, 3, 4, 5]
for i in lista:
    print(i)
```

La variable `i` toma cada uno de los valores de la lista en cada iteración.

## Bucle while
Esta instrucción se utiliza para ejecutar un bloque de código mientras se cumpla una condición. Es decir, el bucle se ejecutará mientras la condición booleana sea verdadera.

```python
while condicion:
    # código a ejecutar
```

Ejemplo:
```python
i = 0
while i < 5:
    print(i)
    i += 1
```

La variable `i` se incrementa en 1 en cada iteración, y el bucle se ejecutará mientras `i` sea menor que 5.

### Ejercicio — Conversor de unidades

En física, una cantidad no es solo un número: es un número **junto con sus dimensiones**. Vamos a crear un conversor de unidades que verifique que las unidades sean compatibles. Para ello, vamos a usar una lista de tuplas que contenga las unidades y su factor de conversión a la unidad fundamental del SI (metro).

`````python
tabla_distancias = [
    ("mm", 0.001),
    ("cm", 0.01),
    ("m", 1.0),
    ("km", 1000.0),
    ("mi", 1609.34),
    ("ft", 0.3048),
    ("in", 0.0254),
]

def buscar_factor(unidad):
    """Busca el factor de conversión de 'unidad' recorriendo la lista de tuplas."""
    for nombre_unidad, factor in tabla_distancias:
        if nombre_unidad == unidad:
            return factor

    raise ValueError(f"Unidad '{unidad}' no reconocida")

def convertir_longitud(valor, unidad_origen, unidad_destino):
    """Convierte 'valor' de unidad_origen a unidad_destino usando la lista de tuplas."""
    factor_origen = buscar_factor(unidad_origen)
    factor_destino = buscar_factor(unidad_destino)
    metros = valor * factor_origen
    return metros / factor_destino

# Pruebas
print(convertir_longitud(5, "mm", "km"))
print(convertir_longitud(12, "in", "cm"))
`````

## Expansión numérica en series de Taylor

$$
    f(x) = \sum_{n=0}^\infty \frac{1}{n!} f^{(n)}(x_0) (x - x_0)^n
$$

Crear una función que expanda $\cos(\theta)$ alrededor de $\theta_0 = 0$ con un número máximo de términos dados por una variable `Nmax`

`````python
from math import factorial

def coeffs(nmax):
    Cs = []

    k = 0
    for i in range(nmax):
        if i%2 == 0:
            Cs.append((-1)**k)
            k += 1
        else:
            Cs.append(0)

    return Cs

Cs = coeffs(100)

def f_approximated(x,nmax):
    return sum([ 1/factorial(n)*Cs[n]*x**n for n in range(nmax) ])
`````

`````python
import numpy as np
import matplotlib.pyplot as plt
`````

`````python
x = np.linspace(-np.pi, np.pi, 100 )
y_real = np.cos(x)

plt.plot(x, y_real, "--", label = "cos(x)", lw=3)
plt.xlabel("x")
plt.ylabel("y")

for nmax in range(3,10):
    y_approx = [ f_approximated(xi,nmax) for xi in x]
    plt.plot(x, y_approx, label = f"nmax = {nmax}")

plt.axis([-np.pi, np.pi, -1,1])
plt.legend()
plt.show()
`````
